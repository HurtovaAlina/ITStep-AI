# Завдання 6
Перевірте чи правильно розраховано загальну вартість покупки. Виведіть ті рядки де є помилка. Виправте дані в таблиці

```python
new_table_with_lower_case['total_calculation'] =  new_table_with_lower_case['unit_price'] * df['units']
table_with_calc = new_table_with_lower_case
table_with_calc
```

```python
mask_wrong = table_with_calc['sale_amt'] != table_with_calc['total_calculation']
mask_wrong
```

```python
table_with_calc[mask_wrong]
```

```python
table_with_calc.loc[mask_wrong, 'sale_amt'] = (
    table_with_calc.loc[mask_wrong, 'total_calculation']
)
table_with_calc
```
